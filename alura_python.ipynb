```python
nome = input(str('Qual o seu nome?: '))
nome = nome.upper()
estado = input(str('Qual o seu estado?: '))

print('====================================')
nota_prova1 = input('Digite a nota da prova 1: ')
nota_prova2 = input('Digite a nota da prova 2: ')
nota_prova3 = input('Digite a nota da prova 3: ')

avg_notas = (float(nota_prova1) + float(nota_prova2) + float(nota_prova3)) / 3
print('====================================')
print(f'A média das notas é: {avg_notas}')
print('====================================')
if avg_notas < 6:
    print('Você foi reprovado')
else:
    print('Você foi aprovado')
    print(f'Continue estudando {nome}')
print('====================================')
if estado == 'RS' or estado == 'SC':
  print('Você é viado')
```

### Exercício - Gerenciando custos de entrega

A Hermex Log, uma empresa de logística especializada em serviços de entrega, está buscando otimizar seus custos operacionais.

Recentemente, a empresa decidiu alugar um novo depósito para armazenar pacotes antes da distribuição.

O aluguel mensal do depósito é de R$ 5.000. Além disso, a empresa gasta R$ 2.000 mensais em combustível e R$ 1.500 em manutenção de veículos.

A equipe de finanças deseja calcular o custo total mensal e, em seguida, dividir esse custo entre as 10 rotas de entrega que a empresa opera, para entender o custo médio por rota.

Como a equipe de finanças pode calcular o custo total mensal e o custo por rota usando conceitos de variáveis e operações matemáticas?

```python
aluguel_deposito = 5000
combustivel = 2000
manutencao_veicular = 1500
qty_rotas = 10
custo_total = aluguel_deposito + combustivel + manutencao_veicular

custo_por_rota = custo_total / qty_rotas

print(f'O custo total mensal é de R$ {custo_total}')
print(f'O custo por rota é de {custo_por_rota}')
```
