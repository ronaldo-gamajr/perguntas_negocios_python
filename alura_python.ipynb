**Desafio 1 — Aprovação de desconto comercial**

Uma empresa permite descontos de acordo com o valor do pedido:

*   Até R$ 500,00:  sem desconto;

*   Acima de R$ 500,00  até R$ 2.000,00: 5% de desconto;

*   Acima de R$ 2.000,00  até R$ 5.000,00: 10% de desconto;

*   Acima de R$ 5.000,00: 15% de desconto.

Crie um programa que solicite o valor do pedido e apresente:

* Valor original;
* Percentual de desconto;
* Valor descontado;
* Valor final do pedido.

Exemplo de saída:

* Valor do pedido: R$ 3.000,00
* Desconto aplicado: 10%
* Valor do desconto: R$ 300,00
* Valor final: R$ 2.700,00

```python
# Entrada
print("============== Dez Company=================\n")
valor_pedido = float(input("Informe o valor do pedido R$: "))

# Processamento
desc_1 = 0.05
desc_2 = 0.10
desc_3 = 0.15

if valor_pedido > 500 and valor_pedido <= 2000:
  desconto = valor_pedido * desc_1
  valor_final = valor_pedido - desconto
  desc_aplicado = desc_1
elif valor_pedido > 2000 and valor_pedido <= 5000:
  desconto = valor_pedido * desc_2
  valor_final = valor_pedido - desconto
  desc_aplicado = desc_2
elif valor_pedido > 5000:
  desconto = valor_pedido * desc_3
  valor_final = valor_pedido - desconto
  desc_aplicado = desc_3
else:
  valor_final = valor_pedido
  desc_aplicado = 0

# Saída
print("\n============== Resumo do Pedido =================\n")
print(f"Valor do pedido: R$ {valor_pedido}")
print(f"% de desconto: {desc_aplicado:.2f}")
print(f"Desconto: R$ {desconto:.2f}\n")
print(f"Valor total da compra: R$ {valor_final}\n")
print("========== Obrigado, volte sempre! ==========")
```

**Desafio 2 — Classificação de prioridade de chamado**

Um Service Desk classifica os chamados com base no impacto e na urgência.

Solicite:

- Impacto: baixo, médio ou alto;
- Urgência: baixa, média ou alta.

Utilize as seguintes regras:

- Impacto alto e urgência alta: prioridade Crítica;
- Impacto alto ou urgência alta: prioridade Alta;
- Impacto médio e urgência média: prioridade Média;
- Demais casos: prioridade Baixa.

Exemplo:

- Impacto: alto
- Urgência: alta
- Prioridade do chamado: Crítica

```python
# Entrada
impacto = input("Qual o impacto? (baixo, médio, alto): ")
urgencia = input("Qual a urgência? (baixa, média, alta): ")

# Processamento
impacto = impacto.capitalize()
urgencia = urgencia.capitalize()

if impacto == "Alto" and urgencia == "Alta":
  prioridade_chamado = "Crítica"
elif impacto == "Alto" and urgencia == "Alta":
  prioridade_chamado = "Alta"
elif impacto == "Médio" and urgencia == "Média":
  prioridade_chamado = "Média"
else:
  prioridade_chamado = "Baixa"

# Saída
print(f"\nImpacto informado: {impacto}")
print(f"Urgência informada: {urgencia}\n")
print(f"Prioridade do chamado: {prioridade_chamado}")
```

**Desafio 3 — Análise de cumprimento de SLA**

Uma empresa possui os seguintes prazos de atendimento:

- Prioridade alta: 4 horas;
- Prioridade média: 8 horas;
- Prioridade baixa: 16 horas.

Solicite:

- Prioridade do chamado;
- Tempo gasto no atendimento, em horas.

O programa deve informar:

- SLA cumprido;
- SLA violado;
- Quantas horas restaram ou quantas horas ultrapassaram o prazo.

Exemplo:

- Prioridade: média
- Tempo de atendimento: 10 horas

- Resultado: SLA violado
- Tempo excedido: 2 horas

```python
# Entrada
print("************* Service Desk ******************\n")
prioridade = input("Qual a prioridade do chamado? (alta, média ou baixa): ")
tempo_atendimento = int(input("Informe o tempo de atendimento do chamado, em horas inteiras: "))

# Processamento
p_baixa = 16
p_media = 8
p_alta = 4

prioridade = prioridade.capitalize()

if prioridade == "Baixa" and tempo_atendimento <= p_baixa:
  sla_status = "Cumprido"
  tempo_excedido = 0
else:
  if prioridade == "Baixa" and tempo_atendimento > p_baixa:
    sla_status = "SLA Violado"
    tempo_excedido = tempo_atendimento - p_baixa
if prioridade == "Média" and tempo_atendimento <= p_media:
  sla_status = "Cumprido"
  tempo_excedido = 0
else:
  if prioridade == "Média" and tempo_atendimento > p_media:
    sla_status = "SLA Violado"
    tempo_excedido = tempo_atendimento - p_media
if prioridade == "Alta" and tempo_atendimento <= p_alta:
  sla_status = "Cumprido"
else:
  if prioridade == "Alta" and tempo_atendimento > p_alta:
    sla_status = "SLA Violado"
    tempo_excedido = tempo_atendimento - p_alta


# Saída
print(f"\nPriodidade: {prioridade} ")
print(f"Tempo de atendimento: {tempo_atendimento}")
print(f"Resuldado: {sla_status}")
print(f"Tempo excedido: {tempo_excedido} h")
```

**Desafio 5 — Cálculo de comissão de vendedor**

Uma empresa paga comissão com base no faturamento mensal:

- Até 10.000,00: 2%;

- De 10.000,01 até R$ 30.000,00: 4%;

- De 30.000,01 até R$ 50.000,00: 6%;

- Acima de 50.000,00: 8%.

Caso o vendedor tenha alcançado a meta de novos clientes, recebe um bônus adicional de R$ 500,00.

Solicite:

- Nome do vendedor;
- Valor total vendido;
- Quantidade de novos clientes;
- Meta de novos clientes.

Apresente:

- Percentual de comissão;
- Valor da comissão;
- Bônus;
- Remuneração variável total.

```python
# Faixas de comissao
comissao = [0.02, 0.04, 0.06, 0.08]
bonus = 500

# Faixas de faturamento
faturamento = [10000, 30000, 50000]

# Entrada
vendedor = input("Qual o nome do vendedor?: ")
vendedor = vendedor.capitalize()
valor_vendido = float(input(f"Quando o {vendedor} vendeu?: "))
qty_clientes = int(input(f"Quantos novos clientes o {vendedor} adicionou em sua carteira?: "))
meta_clientes = int(input(f"Qual a meta de novos clientes?: "))

# Processamento -> Valor comissão
if valor_vendido <= faturamento[0]:
  percent_comissao = comissao[0]
  valor_comissao = valor_vendido * percent_comissao
elif valor_vendido > faturamento[0] and valor_vendido <= faturamento[1]:
  percent_comissao = comissao[1]
  valor_comissao = valor_vendido * percent_comissao
elif valor_vendido > faturamento[1] and valor_vendido < faturamento[2]:
  percent_comissao = comissao[2]
  valor_comissao = valor_vendido * percent_comissao
elif valor_vendido > faturamento[2]:
  percent_comissao = comissao[3]
  valor_comissao = valor_vendido * percent_comissao

# Processamento -> Bônus
if qty_clientes >= meta_clientes:
  remuneracao_total = valor_comissao + bonus
  status_bonus = "Bônus aplicável"
else:
  remuneracao_total = valor_comissao
  status_bonus = "Bônus não aplicável"

# Saída
print(f"\n****************** Resumo do vendedor {vendedor} ***************************\n")
print(f"Percentual de comissão: {percent_comissao:}")
print(f"Valor da comissão: R$ {valor_comissao:}")
print(f"Bônus: R$ {bonus:}")
print(f"Status do bônus: {status_bonus}")
print(f"Remuneração variável total: R$ {remuneracao_total:}")
```

**Desafio 1 — Registro diário de vendas**

Uma loja precisa registrar as vendas realizadas durante o dia.

Solicite continuamente o valor de cada venda. O registro deve ser encerrado quando o usuário digitar 0.

Ao final, apresente:

- Quantidade de vendas;
- Faturamento total;
- Ticket médio;
- Maior venda;
- Menor venda.

Exemplo:

- Digite o valor da venda: 150
- Digite o valor da venda: 300
- Digite o valor da venda: 50
- Digite o valor da venda: 0

Quantidade de vendas: 3
- Faturamento total: RS 500,00
- Ticket médio: 166,67
- Maior venda: 300,00
- Menor venda: 50,00

```python
# Entrada
print("Para encerrar, digite 0")
registro_venda = list()
valor_venda = float(input("Digite o valor da venda R$: "))

# Processamento
while valor_venda != 0 and valor_venda > 0:
  registro_venda.append(valor_venda)
  valor_venda = float(input("Digite o valor da venda R$: "))

# Saida
if len(registro_venda) > 0: # Verifica se há vendas registradas
  # Processamento -> Faturamento
  faturamento_total = sum(registro_venda)
  ticket_medio = faturamento_total / len(registro_venda)
  maior_venda = max(registro_venda)
  menor_venda = min(registro_venda)

  print("\n================ Resumo da Operação =================\n")
  print(f"Faturamento total: R${faturamento_total}")
  print(f"Ticket médio: R${ticket_medio}")
  print(f"Maior venda: R${maior_venda}")
  print(f"Menor venda: R${menor_venda}")
  print(f"Registro das vendas: {registro_venda}")
else:
  print("Nenhuma venda foi registrada.")
```

**Desafio 2 — Controle de estoque**

Uma empresa possui 100 unidades de um produto em estoque.

O programa deve solicitar continuamente uma quantidade para retirada.

Regras:

- A retirada só pode ocorrer se houver saldo suficiente;
- Quantidades negativas não devem ser aceitas;
- Digitar 0 encerra o sistema;
- Quando o estoque chegar a 20 unidades ou menos, exiba um alerta;
- Quando o estoque chegar a zero, encerre o programa automaticamente.

Ao final, apresente:

- Estoque inicial;
- Total retirado;
- Estoque restante;
- Quantidade de retiradas realizadas.

```python
estoque_inicial = 100
estoque = estoque_inicial
total_retirado = 0
quantidade_retiradas = 0

print("Controle de Estoque - Digite 0 para encerrar a operação\n")

while True:
  try:
    qty_retirada = int(input(f"Estoque atual: {estoque}.\nDigite a quantidade a ser retirada: "))

    if qty_retirada == 0:
      print("Operação encerrada pelo usuário.")
      break

    if qty_retirada < 0:
      print("Erro: Quantidades negativas não são aceitas. Tente novamente.")
      continue

    if qty_retirada > estoque:
      print("Erro: Saldo insuficiente em estoque. Tente novamente com uma quantidade menor.")
      continue

    # Processamento da retirada
    estoque -= qty_retirada
    total_retirado += qty_retirada
    quantidade_retiradas += 1

    if estoque <= 20 and estoque > 0:
      print("Alerta: estoque baixo!")
      decisao = input("Deseja continuar operando? (S/N): ").strip().upper()
      if decisao == 'N':
        print("Operação encerrada devido ao estoque baixo.")
        break

    if estoque <= 0:
      print("Estoque zerado! Programa encerrado automaticamente.")
      break

  except ValueError:
    print("Entrada inválida. Por favor, digite um número inteiro.")

# Saída final
print("\n================ Resumo do Estoque ================\n")
print(f"Estoque inicial: {estoque_inicial}")
print(f"Total retirado: {total_retirado}")
print(f"Estoque restante: {estoque}")
print(f"Quantidade de retiradas realizadas: {quantidade_retiradas}")
print("\n===================================================")
```
